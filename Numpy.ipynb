```python
# Q. Write a Python Program to calculate mode without using inbuilt function 
    # input :- 5,1,1,1,2,3
    # output :- mode - 1
# Ans.

x = int(input("enter length of items"))
y = []

for i in range(x):
    z = int(input(f"number {i} ="))
    y.append(z)
print(x)
print(y)
mode = y[0]
max_count = 0
for i in y:
    count = 0
    for j in y:
        if i == j:
            count += 1
    if count > max_count:
        max_count = count
        mode = i
print("mode -", mode)
```

```python
# Write a python program to caculate the median of a group of numbers 
# Input :- 
            # 5
            # 12
            # 10
            # 16
            # 14
            # 18
# Output :-
            # 14

# Ans. 
x = int(input("enter length of items"))
y = []
for i in range(x):
    z = int(input(f"number {i} ="))
    y.append(z)
print(x)
print(y)
y.sort()
print(y)
# if number of element is even 
if (x%2 == 0): # even > True
    print("if part run", (x+1)//2)
    right = y[(x+1)//2]
    print(right)
    left = y[(x-1)//2]
    print(left)
    median = (right+left)/2
    print(median)
else:
    print(y[(x//2)])
```

```python
# flatten | create copy then works
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)

# 2D -> 1D
flat_arr = arr.flatten()
print(flat_arr)
flat_arr[0] = 100
print(flat_arr)
print(arr)

# 3D -> 1D
arr1 = np.arange(24).reshape(3,4,2)
print(arr1)
flat_arr1 = arr1.flatten()
print(flat_arr1)
```

```python
# Ravel | work on original array
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)

# 2D -> 1D
rav_arr = arr.ravel()
rav_arr[0] = 100
print(rav_arr)
print("Updated array :- ",arr)

# 3D -> 1D
arr3 = np.arange(24).reshape(3,4,2)
print(arr3)
rav_arr3 = arr3.ravel()
print(rav_arr3)
```

```python
# transpose
import numpy as np
arr = np.arange(12).reshape(3,4)
print(arr)
trans_arr = arr.transpose() # Alternative :- T
print(trans_arr) # row ko cols me change karta h

# 3D 
arr3 = np.arange(24).reshape(3,4,2)
print(arr3)
trans_arr3 = arr3.T
trans_arr3
```

```python
# adding and removing data

import numpy as np

numbers = np.array([
    [5, 15, 25],
    [35, 45, 55],
    [65, 75, 85]
])

print(numbers)


# 1. concatenate
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

new_data = np.concatenate((a, b))
print(new_data)


# 2. vstack - add new row
row = np.array([[95, 105, 115]])

numbers = np.vstack((numbers, row))
print(numbers)


# 3. hstack - add new column
column = np.array([
    [100],
    [200],
    [300],
    [400]
])

numbers = np.hstack((numbers, column))
print(numbers)


# 4. delete - remove row
numbers = np.delete(numbers, 1, axis=0)
print(numbers)
```
